# Software di un negozio di prodotti vegani

Questo progetto consiste nel realizzare un software per la gestione di un negozio di prodotti vegani.

**Metodo di sviluppo del progetto:**

Applicazione della Programmazione ad Oggetti (OOP) per rendere il software riutilizzabile e più facile da mantenere.

Gestione delle eccezioni e verifica degli input inseriti dall'utente.

Salvataggio delle informazioni inserite dall'utente in un file JSON, al fine di essere mantenute tra diverse esecuzioni del programma.

Le funzionalità del software saranno gestite da apposite funzioni, e permetteranno all'utente di:

- Registrare nuovi prodotti indicando nome, quantità, prezzo di vendita e prezzo di acquisto.
- Elencare tutti i prodotti presenti.
- Registrare le vendite effettuate.
- Mostrare i profitti lordi e netti.
- Mostrare un menù di aiuto con tutti i comandi disponibili.

L’interfaccia utente sarà basata su riga di comando: l'utente potrà consultare il menù principale con opzioni descritte in maniera semplice.

1. **Installazione moduli e librerie**

- json = modulo JSON per operazioni su file json. Viene utilizzato nel software per caricare, visualizzare, scrivere e salvare i dati del registro di magazzino.
- os = modulo Operating System per interagire con il sistema operativo. Viene utilizzato nel software per verificare l'esistenza del file json all'avvio del programma.

2. **Creazione delle classi e metodi**

- classe Product: rappresenta un singolo prodotto. Prende come attributi privati nome, quantità, prezzo di acquisto e prezzo di vendita.

- classe Register: rappresenta il registro di magazzino di tutti i prodotti. Prende come attributi privati il registro di magazzino, il file sul quale il registro viene salvato, la lista di profitti e la lista di costi dei ogni operazione effettuata sul magazzino. Prevede inoltre tutti i metodi per effetturare tutte le operazioni di movimentazione dei prodotti di magazzino.

3. **Funzionalità**:
- attraverso la funzione di input interattiva di aggiunta di un nuovo prodotto "input_add_product", l'utente può aggiungere un nuovo prodotto nel magazzino o aggiornare la quantità e i prezzi di un prodotto già esistente.
- attraverso la funzione di input interattiva di vendita di un prodotto "input_sell_product", l'utente può registrare una vendita di uno o più prodotti presenti nel magazzino e aggiornare la quantità disponibile dopo la vendita.

4. **Interfaccia**:
attraverso la funzione main() viene avviato il software di gestione di magazzino mediante il quale l'utente può selezionare una delle operazioni disponibili dal menù:

  1. aggiungi: registrare nuovi prodotti, con nome, quantità, prezzo di vendita e prezzo di acquisto.
  2. elenca: elencare tutti i prodotti presenti in magazzino.
  3. vendita: registrare le vendite effettuate per ogni prodotto e visualizzare lo scontrino.
  4. profitti: mostrare i profitti lordi e netti totali cumulati.
  5. aiuto: visualizzare il menù con la descrizione dei comandi.
  6. chiudi: terminare l'esecuzione del programma.




```python

import json
import os

"""
    La classe Product gestisce le informazioni relative a un prodotto del magazzino di prodotti vegani.

    Attributi di istanza:
    _name (str): Nome del prodotto. Default: None.
    _quantity (int): Quantità disponibile in magazzino. Default: None.
    _buy_price (float): Prezzo di acquisto del prodotto. Default: None.
    _sell_price (float): Prezzo di vendita del prodotto. Default: None.

    Metodi:
    __init__(self, name=None, quantity=None, buy_price=None, sell_price=None):
    Costruttore della classe. Inizializza gli attributi del prodotto.

    Parametri:
    name (str, opzionale): Nome del prodotto. Default: None.
    quantity (int, opzionale): Quantità disponibile in magazzino. Default: None.
    buy_price (float, opzionale): Prezzo di acquisto del prodotto. Default: None.
    sell_price (float, opzionale): Prezzo di vendita del prodotto. Default: None.
"""

class Product:

  def __init__(self, name=None, quantity=None, buy_price=None, sell_price=None):
    self._name = name
    self._quantity = quantity
    self._buy_price = buy_price
    self._sell_price = sell_price

  '''
  Lettura degli attributi privati della classe con metodo get:

  get_name(self): ritorna il nome del prodotto.
  get_get_quantity(self): ritorna la quantità del prodotto selezionato.
  get_get_buy_price(self): ritorna il prezzo di acquisto del prodotto selezionato.
  get_get_sell_price(self): ritorna il prezzo di vendita del prodotto selezionato.

  '''

  def get_name(self):
    return self._name

  def get_quantity(self):
    return self._quantity

  def get_buy_price(self):
    return self._buy_price

  def get_sell_price(self):
    return self._sell_price


  """
    La classe Register gestisce il registro delle operazioni di movimentazione del magazzino.
    Fornisce funzionalità per salvare e caricare i dati in/da un file JSON.

    Attributi di istanza:
    _warehouse_reg (dict): Dizionario che rappresenta il registro del magazzino, contenente i prodotti e le relative informazioni.
      Se non specificato, viene inizializzato come un dizionario vuoto.
    _filename (str): Nome del file JSON in cui viene salvato il registro del magazzino. Default: "warehouse_register.json".
    _profits (list): Lista delle transazioni di vendita per il calcolo dei profitti. Inizializzata come lista vuota.
    _costs (list): Lista delle transazioni di acquisto per il calcolo dei costi. Inizializzata come lista vuota.

    Metodi:
    __init__(self, warehouse_reg=None, filename="warehouse_register.json"):
    Costruttore della classe. Inizializza il registro e carica i dati dal file JSON se disponibile.

    Parametri:
      - warehouse_reg (dict, opzionale): Registro iniziale dei prodotti. Se None, viene inizializzato come un dizionario vuoto.
      - filename (str, opzionale): Nome del file JSON per il salvataggio/caricamento del registro. Default: "warehouse_register.json".

    open_warehouse_reg_json():
    Metodo chiamato nel costruttore per caricare il registro dal file JSON, se esiste.
  """

class Register:
  def __init__(self, warehouse_reg=None, filename="warehouse_register.json"):
    if warehouse_reg is None:
      warehouse_reg = {}
    self._warehouse_reg = warehouse_reg
    self._filename = filename
    self.open_warehouse_reg_json()
    self._profits=[]
    self._costs=[]

  """
  Lettura degli attributi privati della classe con metodo get:
  get_warehouse_reg(self): ritorna il registro del magazzino.
  """

  def get_warehouse_reg(self):
    return self._warehouse_reg

  """
  def open_warehouse_reg_json(self):
  Metodo chiamato nel costruttore per caricare il registro dal file JSON, se esiste.
  Se il registro del magazzino esiste, il file viene caricato
  Se il registro del magazzino non esiste, viene creato e inizializzato un dizionario vuoto.
  Se il programma riporta un eccezione nel caso in cui il file JSON sia corrotto o non leggibile, l'eccezione viene gestita.
  In tal caso verrà creato e inizializzato un dizionario vuoto.

  """

  def open_warehouse_reg_json(self):
    try:
      if os.path.exists(self._filename):
        with open(self._filename, "r") as json_file:
          self._warehouse_reg = json.load(json_file)
        print("\nIl magazzino contiene prodotti già registrati\n")
      else:
        self._warehouse_reg = {}
        with open(self._filename, "w") as json_file:
            json.dump(self._warehouse_reg, json_file, indent=4)
            print("\nLa lista dei prodotti in magazzino è vuota!\n")

    except json.JSONDecodeError:
      print("\nNon è possibile aprire il file, ne verrà creato uno nuovo.\n")
      self._warehouse_reg = {}
      with open(self._filename, "w") as json_file:
        json.dump(self._warehouse_reg, json_file, indent=4)

  """
  def save_product_json(self):
  Metodo chiamato dalle funzioni che gestiscono le operazioni di movimentazione del magazzino per salvare nel registro del magazzino acquisti e vendite di ogni prodotto.
  """

  def save_product_json(self):
      with open(self._filename, "w") as json_file:
        json.dump(self._warehouse_reg, json_file, indent=4)
  """
  add_product(self, product):

   Registra l'aggiunta di un nuovo prodotto (e relative informazioni) nell magazzino o l'aumento della quantità di un prodotto esistente a seguito di un acquisto.

    Parametri:
    product (Product): oggetto della classe Product, ottenuto dalla funzione input_add_product e passato a questa funzione per essere aggiunto al magazzino.

    Funzionamento:
    Utilizza i metodi `get` definiti nella classe Product per ottenere le informazioni del prodotto.
    Controlla se il prodotto è già presente nel magazzino.

    1. Se il prodotto è già presente:
    La quantità aggiuntiva viene registrata nella lista `self._costs` per mantenere la cronologia degli ordini.
    Il registro del magazzino viene aggiornato con le nuove unità acquistate.

    2. Se il prodotto non è presente:
    Il prodotto viene aggiunto al registro del magazzino.
    La lista `self._costs` viene aggiornata con l'acquisto del nuovo prodotto.

  """

  def add_product(self, product):
      product_name = product.get_name()
      product_quantity = product.get_quantity()
      product_buy_price = product.get_buy_price()

      if product_name in self._warehouse_reg:
            additional_quantity = product_quantity - self._warehouse_reg[product_name]['quantity']
            if additional_quantity > 0:
                self._costs.append({
                    "prodotto": product_name,
                    "quantità": additional_quantity,
                    "prezzo": product_buy_price
                })
            self._warehouse_reg[product_name]['quantity'] += product.get_quantity()

      else:
            self._costs.append({
                "prodotto": product_name,
                "quantità": product_quantity,
                "prezzo": product_buy_price
            })

            self._warehouse_reg[product_name] = {
                'quantity': product.get_quantity(),
                'buy_price': product.get_buy_price(),
                'sell_price': product.get_sell_price()
            }

      self.save_product_json()
  """
  sell_product(self, product):

   Registra la vendita di un prodotto aggiornando la quantità disponibile nel magazzino.

    Parametri:
    product (Product): oggetto della classe Product, ottenuto dalla funzione `input_sell_product`,
                       che rappresenta il prodotto da vendere.

    Funzionamento:
    Controlla che `product` non sia None.
    Aggiorna la quantità disponibile nel magazzino.
    Salva il registro aggiornato in formato JSON.
  """

  def sell_product(self, product):
      if product is None:
        return

      product_name = product.get_name()
      self._warehouse_reg[product_name]['quantity'] = product.get_quantity()
      self.save_product_json()

  """
  profits(self):

  Calcola il profitto lordo complessivo accedendo alla lista cumulata di tutte le vendite effettuate.
  Il profitto lordo è definito come la somma dei ricavi generati dalla vendita di ogni prodotto, pari a prezzo di vendita * quantità venduta.
  """

  def profits(self):
        profitto_lordo = 0
        for item in self._profits:
            nome_prodotto = item["prodotto"]
            quantità_prodotto = item["quantità"]
            prezzo_unitario = item["prezzo"]

            ricavi = quantità_prodotto * prezzo_unitario
            profitto_lordo += ricavi

        print(f"Profitto Lordo: {profitto_lordo}\n")

  """
  costs(self):

  Calcola i costi complessivi accedendo alla lista cumulata di tutti gli acquisti effettuati.
  I costi totali sono definiti come la somma dei costi sostenuti per l'acquisto di ogni prodotto, pari a prezzo di acquisto * quantità acquistata.
  Infine verifica se si sono registrati anche profitti dalle vendita in self._profits e se si, calcola il profitto netto.
  Il profitto netto è pari alla sommatoria di tutte le quantità vendute e i prezzi di vendita per ogni prodotto (profitto lordo), meno i costi totali.
  """

  def costs (self):
      costi_totali=0

      for item in self._costs:
            nome_prodotto = item["prodotto"]
            quantità_prodotto = item["quantità"]
            prezzo_unitario = item["prezzo"]
            costi = quantità_prodotto * prezzo_unitario
            costi_totali += costi

      print(f"Costi Totali: {costi_totali}\n")

      if self._profits:
          profitto_netto = sum(item["quantità"] * item["prezzo"] for item in self._profits) - costi_totali
          print(f"Profitto Netto: {profitto_netto}\n")

  """
  transaction (self, transaction_items):

  Calcola l'importo totale venduto della singola operazione di vendita, riportando i dati di vendita (prodotto, quantità, prezzo e totale venduto) registrati all'interno del ciclo.
  Se si effettua una successiva operazione di vendita, la lista transaction viene azzerata, in modo che il riepilogo della transazione non dia un risultato cumulato di tutte le operazioni di vendita.
  """

  def transaction (self, transaction_items):
    tot_venduto =0
    for item in transaction_items:
        nome_prodotto = item["prodotto"]
        quantità_prodotto = item ["quantità"]
        prezzo_unitario = item["prezzo"]

        prezzo_totale = quantità_prodotto*prezzo_unitario

        tot_venduto +=prezzo_totale

        print(f"Prodotto: {nome_prodotto}, Quantità: {quantità_prodotto}, Prezzo unitario: {prezzo_unitario}, Prezzo totale: {prezzo_totale}\n")
        print(f"Totale venduto: {tot_venduto}\n")
  """
  print_warehouse(self):
  Elenca i prodotti presenti nel registro del magazzino (se ve ne sono) riepilogando prodotto, quantità e prezzo di ognuno.
  Se nel magazzino non sono presenti prodotti fornisce un messaggio che informa che il magazzino è vuoto.

  """

  def print_warehouse(self):
      if not self._warehouse_reg:
          print("\nIl magazzino è vuoto.\n")
      else:
          print("\nPRODOTTO QUANTITA' PREZZO\n")
          for prodotto, dati in self._warehouse_reg.items():
            quantità = dati.get('quantity', 0)
            prezzo = dati.get('sell_price', 0)
            print(f"{prodotto} {quantità} €{prezzo}\n")

  """
    Gestisce l'input dell'utente per aggiungere un nuovo prodotto al magazzino.

    Questa funzione richiede all'utente di inserire le informazioni necessarie per un nuovo prodotto (nome, quantità, prezzo di acquisto e prezzo di vendita) non presente in magazzino.
    Se il prodotto esiste già, chiede solo la quantità aggiuntiva se il prezzo di vendita e il prezzo di acquisto sono presenti nel registro, altrimenti chiede di aggiungere anche tali informazioni.
    Gestisce la validazione dell'input e mostra messaggi di errore appropriati.

    Argomenti:
    warehouse_reg (Register): L'istanza del registro del magazzino.

    Risultato:
    Product: restituisce un nuovo oggetto Product con le informazioni inserite dall'utente.

    Gestione delle eccezioni:
    AssertionError: Se l'input dell'utente non è valido.
    ValueError: Se l'input numerico non è valido.
  """


def input_add_product(warehouse_reg):
  while True:
    try:
        add_new_product = input("Nome del prodotto: \n").strip().title()
        assert add_new_product != "", "Il nome prodotto non può essere vuoto. Riprova.\n"

        if add_new_product in warehouse_reg.get_warehouse_reg():
          print(f"\nProdotto {add_new_product} già presente in magazzino.\n")
          while True:
            try:
                quantity_to_add = input("Quantità: \n").strip().title()
                assert quantity_to_add.isdigit(), "La quantità acquistata deve essere un numero intero. Riprova.\n"
                quantity_to_add = int(quantity_to_add)
                print(f"quantità aggiunta a {add_new_product} pari a {quantity_to_add}.\n")
                break
            except AssertionError as e:
              print(e)

          existing_product = warehouse_reg.get_warehouse_reg()[add_new_product]

          if 'buy_price' not in existing_product or 'sell_price' not in existing_product:
                    print(f"Il prodotto {add_new_product} esistente non ha un prezzo di acquisto o di vendita registrato.")
                    while True:
                        try:
                            buy_price = float(input("Prezzo di acquisto: \n"))
                            sell_price = float(input("Prezzo di vendita: \n"))
                            break
                        except ValueError:
                            print("\nInput non valido. Inserisci numeri validi.\n")

                    existing_product['buy_price'] = buy_price
                    existing_product['sell_price'] = sell_price


          return Product(name=add_new_product,
                      quantity=existing_product['quantity'] + quantity_to_add,
                      buy_price=existing_product['buy_price'],
                      sell_price=existing_product['sell_price'])

          warehouse_reg._costs.append({"prodotto": add_new_product, "quantità": quantity_to_add, "prezzo": buy_price})

        else:
          while True:
            try:
              buy_price = float(input("Prezzo di acquisto: \n"))
              sell_price = float(input("Prezzo di vendita: \n"))
              quantity = int(input("Quantità: \n"))
              return Product(name=add_new_product, quantity=quantity,
                                      buy_price=buy_price, sell_price=sell_price)
            except ValueError:
              print("\nInput non valido. Inserisci numeri validi.\n")


    except AssertionError as e:
      print("\nSi è verificato un errore:", e)

  """
    Gestisce l'input dell'utente per aggiungere la vendita di un prodotto.

    Questa funzione richiede all'utente di inserire le informazioni necessarie per vendere un prodotto presente in magazzino (nome e quantità)
    Se il prodotto non è presente in magazzino, lo segnala e chiede all'utente di riprovare.
    Se il prodotto è presente in magazzino, registra la vendita e chiede all'utente se vuole continuare aggiungendo un altro prodotto.
    Quando l'utente decide di terminare il processo di vendita, viene visualizzato il riepilogo della vendita con le informazioni sui prodotti venduti e il prezzo totale di vendita.
    Gestiscela validazione dell'input e mostra messaggi di errore appropriati.

    Argomenti:
    warehouse_reg (Register): L'istanza del registro del magazzino.
    transaction: la lista che registra le movimentazioni del magazzino nel ciclo di vendita per riepilogare il totale venduto a fine operazione.

    Risultato:
    Product: restituisce un nuovo oggetto Product con le informazioni inserite dall'utente.

    Gestione delle eccezioni:
    AssertionError: Se l'input dell'utente non è valido.
  """

def input_sell_product(warehouse_reg, transaction):
  while True:
    try:
      sell_product = input("Nome del prodotto da vendere: \n").strip().title()
      assert sell_product != "", "\nIl nome prodotto non può essere vuoto. Riprova.\n"

      if not sell_product in warehouse_reg.get_warehouse_reg():
          print(f"Prodotto {sell_product} non presente in magazzino.\n")

          while True:
            try_new_item = input(f"Provare con un altro prodotto? [si/no]\n").strip().lower()
            if try_new_item == "si":
              break
            elif try_new_item == "no":
              return None
            else:
              print("\nRisposta non valida! Inserisci 'si' o 'no'.\n")

      else:
        while True:

          try:
                quantity_to_sell = input("Quantità: \n").strip().title()
                assert quantity_to_sell.isdigit(), "\nLa quantità acquistata deve essere un numero intero. Riprova.\n"
                quantity_to_sell = int(quantity_to_sell)
                existing_product = warehouse_reg.get_warehouse_reg()[sell_product]
                sell_price = existing_product['sell_price']
                if quantity_to_sell > existing_product['quantity']:
                  print(f"stai vendendo una quantità di {sell_product} non presente in magazzino\n")
                  return None
                existing_product['quantity'] -= quantity_to_sell
                print(f"Vendita Registrata\n {quantity_to_sell}X {sell_product}: {sell_price}.\n")
                warehouse_reg.save_product_json()

                transaction.append({"prodotto":sell_product, "quantità":quantity_to_sell, "prezzo":sell_price})
                warehouse_reg._profits.append({"prodotto": sell_product, "quantità": quantity_to_sell, "prezzo": sell_price})

                while True:
                  next_item = input(f"Inserire un altro prodotto? [si/no]\n").strip().lower()
                  if next_item == "si":
                     break
                  elif next_item =="no":
                    return Product(name=sell_product,
                      quantity=existing_product['quantity'],
                      buy_price=existing_product['buy_price'],
                      sell_price=existing_product['sell_price'])
                  else:
                   print("\nRisposta non valida! Inserisci 'si' o 'no'.\n")

                break

          except AssertionError as e:
              print(e)

    except AssertionError as e:
      print("\nSi è verificato un errore:", e)


  """
    Funzione principale del programma di gestione magazzino.

    Inizializza il sistema, crea un'istanza del registro del magazzino
    e gestisce il ciclo principale del programma, presentando un menu
    all'utente e chiamando le funzioni appropriate in base alle scelte
    dell'utente.

  """
def main():


  warehouse_reg = Register()
  profits=[]

  try:
      cmd = None
      while cmd!="chiudi":
        cmd = input("Ciao! Scegli quale operazione eseguire:\n"
                  "1. aggiungi\n"
                  "2. elenca\n"
                  "3. vendita\n"
                  "4. profitti\n"
                  "5. aiuto\n"
                  "6. chiudi\n").strip()


        assert cmd != "", "Inserisci il nome dell'operazione scelta.\n"

        if cmd =="aggiungi":
          print("\nHai richiesto l'operazione 1: Registrare un nuovo prodotto\n")
          new_product = input_add_product(warehouse_reg)
          warehouse_reg.add_product(new_product)

        elif cmd =="elenca":
          print("\nHai richiesto l'operazione 2: Elencare i prodotti in magazzino\n")
          warehouse_reg.print_warehouse()

        elif cmd =="vendita":
          transaction= []
          print("\nHai richiesto l'operazione 3: Registrare una vendita\n")
          sell_product = input_sell_product(warehouse_reg, transaction)
          warehouse_reg.sell_product(sell_product)
          warehouse_reg.transaction(transaction)

        elif cmd =="profitti":
          warehouse_reg.profits()
          warehouse_reg.costs()

        elif cmd =="aiuto":
          print("1. aggiungi: registrare nuovi prodotti, con nome, quantità, prezzo di vendita e prezzo di acquisto.\n"
                  "2. elenca: elencare tutti i prodotti presenti in magazzino.\n"
                  "3. vendita: registrare le vendite effettuate per ogni prodotto e visualizzare lo scontrino.\n"
                  "4. profitti: mostrare i profitti lordi e netti totali cumulati.\n"
                  "5. aiuto: visualizzare il menù con la descrizione dei comandi\n"
                  "6. chiudi: terminare l'esecuzione del programma.\n")

        elif cmd =="chiudi":
          print("bye bye\n")
        else:
          print("il comando che hai inserito non è valido, riprova!\n")

  except AssertionError as e:
    print(e)

  print(f"Fine\n.")

main()
```
